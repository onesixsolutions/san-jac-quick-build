# Cortex Search — Local Dev Notebook

Test Cortex Search queries locally via the Snowflake Python connector.

```python
import os, json
from dotenv import load_dotenv
import snowflake.connector

load_dotenv('../.env')

conn = snowflake.connector.connect(
    account=os.getenv('SNOWFLAKE_ACCOUNT'),
    user=os.getenv('SNOWFLAKE_USER'),
    password=os.getenv('SNOWFLAKE_PASSWORD'),
    role=os.getenv('SNOWFLAKE_ROLE'),
    warehouse=os.getenv('SNOWFLAKE_WAREHOUSE'),
    database=os.getenv('SNOWFLAKE_DATABASE'),
    schema=os.getenv('SNOWFLAKE_SCHEMA'),
)
print('Connected:', conn.account)
```

```python
query = 'What is the financial aid appeal process?'
cur = conn.cursor()
cur.execute("""
    SELECT SNOWFLAKE.CORTEX.SEARCH_PREVIEW(
        'SAN_JAC_DEMO.CORTEX.SAN_JAC_DOCS_SEARCH',
        %s,
        {'limit': 3}
    ) AS response
""", (query,))
response = json.loads(cur.fetchone()[0])
for r in response.get('results', []):
    print(f"--- {r.get('file_name')} ---")
    print(r.get('content', '')[:500])
    print()
```
